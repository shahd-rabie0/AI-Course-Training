### Recursion in Python ###
#### Definiation : ####
- Recursion is when a function Calls itself directly or indirectly to solve a problem .
- it is usually used to break down big problems into smaller , similar sub-problems.   

### Factorial Function ###

```python
# function to calculate factorial 
def factorial (n):
    if n==0 or n==1 : #Base Case
        return 1
    else:
        return n * factorial (n-1) #Recursive Call

print (factorial(5))   #Output : 120 
```

### Filter Function ###

```python

#filter(function,iterable)
n=[11,22,33,44,55]
def check_even(x):
    return x%2==0
filterdlist=[]

for i in n:
    if check_even(i):
        filterdlist.append(i)

 
print(filterdlist)  
```

```python
n=[11,22,33,44,55]
#Using lambda inside filter
filterdlist=list(filter(lambda x :x%2==0,n))
print(filterdlist)
```
