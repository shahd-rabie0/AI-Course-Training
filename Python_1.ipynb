## 3. Python Built-in Data Types

### 3.1 Text Type
* **`str` (String):** Used to store text data.
  * *Example:* `name = "Shahd"`

### 3.2 Numeric Types
* **`int` (Integer):** Used for whole numbers (positive or negative without decimals).
  * *Example:* `age = 22`
* **`float` (Floating point):** Used for numbers with decimals.
  * *Example:* `pi = 3.14`

### 3.3 Sequence Types
* **`list` (List):** Ordered, (changeable) collection of items, enclosed in square brackets.
  * *Example:* `fruits = ["apple", "banana", "cherry"]`
* **`tuple` (Tuple):** Ordered, (unchangeable) collection of items, enclosed in parentheses.
  * *Example:* `coordinates = (10.5, 20.2)`
* **`range` (Range):** Represents a sequence of numbers, commonly used in loops.
  * *Example:* `numbers = range(0, 10)`

### 3.4 Mapping Type
* **`dict` (Dictionary):** Stores data in key-value pairs, mutable and unordered (ordered in Python 3.7+).
  * *Example:* `student = {"name": "Shahd", "age": 22}`

### 3.5 Set Types
* **`set` (Set):** Unordered collection of unique items (no duplicates), mutable.
  * *Example:* `unique_numbers = {1, 2, 3, 3}` *(Results in `{1, 2, 3}`)*

### 3.6 Boolean Type
* **`bool` (Boolean):** Represents truth values, either `True` or `False`.
  * *Example:* `is_active = True`
