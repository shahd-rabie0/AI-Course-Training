```python
x=[1,2,4,5,6]
y=[2,4,8,10,12]
x2=[1,4,16,25,16]
plt.plot(x,y,"r*--",label="equation 1")
plt.plot(x,x2,"b>--",label="equation 2")

plt.title("A linear graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.xticks(np.arange(0,10,1)) # the range of x 
plt.yticks(np.arange(0,30,5))



plt.legend(loc="lower right")
#plt.savefig("graph.svg",dpi=500)
plt.show()
```

```python
x=[1,2,4,5,6]
y=[2,4,8,10,12]
x2=[1,4,16,25,16]
plt.plot(x,y,"r*--",label="equation 1")
plt.plot(x,x2,"b>--",label="equation 2")
plt.plot(x2,y,"g>-",label="equation 3")

plt.title("A linear graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.rcParams["figure.figsize"]=(10,15)

plt.legend()
#plt.savefig("graph.svg",dpi=500)
plt.show()
```

```python
x=[1,2,4,5,6]
y=[2,4,8,10,12]
x2=[1,4,16,25,16]
plt.plot(x,y,"r*--",label="equation 1")
plt.plot(x,x2,"b>--",label="equation 2")
plt.plot(x2,y,"g>-",label="equation 3")

plt.title("A linear graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

axis=plt.gca()#get current axis
axis.spines["top"].set_visible(False)
axis.spines["right"].set_visible(False)

plt.rcParams["figure.figsize"]=(5,5)

plt.legend()
#plt.savefig("graph.svg",dpi=500)
plt.show()
```

Bar chart plots

```python
import matplotlib.pyplot as plt
import numpy as np
labels=["A","B","C"]
values=[1,2,4]
plt.figure(figsize=(6,4))
plt.bar(labels,values,color="#800d3d")
#plt.xticks(rotation=30)
plt.yticks(np.arange(0,5,1))
plt.show()
```

```python
labels=["A","B","C"]
values=[1,2,4]
plt.figure(figsize=(6,4))
plt.bar(labels,values,color="#b82798")
plt.yticks(np.arange(0,5,1))

axis=plt.gca()
axis.spines["top"].set_visible(False)
axis.spines["right"].set_visible(False)

plt.show()

```

```python
labels=["A","B","C"]
values=[1,2,4]
plt.figure(figsize=(6,4))
plt.bar(labels,values,color="#800d3d",align="edge",width=0.5)
plt.yticks(np.arange(0,5,1))
plt.show()
```

```python
labels=["A","B","C"]
values=[1,2,4]
plt.figure(figsize=(6,4))
plt.bar(labels,values,color="#ff0505",width=0.2)
plt.yticks(np.arange(0,5,1))

axis=plt.gca()
axis.spines["top"].set_visible(False)
axis.spines["right"].set_visible(False)

plt.show()
```

```python
labels=["A","B","C"]
values=[4,2,4]

bars=plt.bar(labels,values,color="#0558ff")
bars[0].set_hatch("/")
bars[1].set_hatch("*")
bars[2].set_hatch("o")
plt.show()

```

```python
#to make each one have an unique color
labels=["A","B","C"]
values=[4,2,4]
bar_colors=["red","green","blue"]
bars=plt.bar(labels,values,color=bar_colors)

plt.show
```

```python
plt.figure(figsize=(8,6))
company=["GOOGLE","Amazon","MSFT","FB"]#x_axis
revenue=[90,136,85,27] #Y_axis
profit=[40,7,34,12]#secondry y_axis
xpos=np.arange(len(company))

plt.bar(company,revenue,width=0.4,label="Revenue")
plt.bar(xpos-0.4,profit,width=0.4,label="Profit")

plt.xticks(xpos,company)
plt.ylabel("Revenue(Bln)")
plt.title("US Technology Stocks")

plt.legend()
plt.show()
```

Pie chart plots

```python
exp_vals=[1400,600,300,300,250]
exp_labels=["home rent","food","internet/phone","car","other"]


plt.pie(exp_vals,labels=exp_labels)
plt.show()
```

```python
exp_vals=[1400,600,300,300,250]
exp_labels=["home rent","food","internet/phone","car","other"]


plt.pie(exp_vals,labels=exp_labels,autopct="%1.1f%%")
plt.show()
```

```python
exp_vals=[1400,600,300,300,250]
exp_labels=["home rent","food","internet/phone","car","other"]


plt.pie(exp_vals,labels=exp_labels,autopct="%0.4f%%")
plt.show()
```

```python
exp_vals=[1400,600,300,300,250]
exp_labels=["home rent","food","internet/phone","car","other"]


plt.pie(exp_vals,labels=exp_labels,autopct="%1.1f%%",explode=[0,0,0.2,0,0])
plt.show()
```
