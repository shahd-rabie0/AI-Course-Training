## What is OOP?
**OOP** stands for **Object-Oriented Programming**.
- Python is an object-oriented language, allowing you to structure your code using classes and objects for better organization and reusability.
___________________________________________________________________________________

## Advantages of OOP ##
- Provides a clear structure to programs
- Makes code easier to maintain, reuse, and debug
- Helps keep your code DRY (Don't Repeat Yourself)
- Allows you to build reusable applications with less code
- **Tip**: The DRY principle means you should avoid writing the same code more than once. Move repeated code into functions or classes and reuse it.

# Encapsulation

## 1. What is it?

Collect all the data and the methods that work on it in **one class**, and prevent direct editing from outside.

```java
class ClassName {
    // data members (state)
    // +
    // methods (behaviour)
}
```

**Black box idea:** think of a TV. You press buttons (inputs) and get a picture (outputs). You don't touch the circuits inside.

```
Input  ->  [ Black Box ]  ->  Output
           (hidden inside)
```

---

## 2. Why do we need it?

- It helps the programmer **control** how the data is changed.
- Direct access is **dangerous**, because someone else on the team may be using the same variable in another part of the code.
- It helps **keep track of dependencies** between variables.
- Code becomes **easier to edit and test** later.

---

## 3. Setters and Getters

| | Purpose |
|---|---|
| **Getter** | Read the value only (see it) |
| **Setter** | Change the value (with control) |

**Not every variable needs a setter.** For example, a `color` attribute only needs a getter, because we never need to change it.

**Example: moving a chess piece.** To move a piece, we first need its current position (`row`, `col`), so we use a getter to read it, then a method to move it.

---

## 4. Example: Game Player

```java
class Player {
    private int currentHealth;
    private int maxHealth;
    private int level;
}
```

### Problem
When the player levels up, `maxHealth` increases and `currentHealth` must also be updated. If we edit each variable by hand every time, we can forget a step.

### Solution: one setter that does all the steps

```java
public void setHealthOnLevelUp() {
    maxHealth += 500;
    currentHealth = maxHealth;
}
```

---

## 5. Validation

While walking, the player collects gems and their health increases:

| Gem | Health |
|---|---|
| Green | +300 |
| Red | +1000 |
| Gold | +500 |

### Without a check

```
level = 2
maxHealth = 700
gem = +1000   (red)
```

`currentHealth` can become more than `maxHealth`, which **breaks the game rules**.

### With validation in the setter

```java
public void addHealth(int gem) {
    currentHealth += gem;
    if (currentHealth > maxHealth) {
        currentHealth = maxHealth;   // never exceed the max
    }
}
```

---

## 6. Example: Car (easy to edit)

```java
class Car {
    private String gasType = "diesel";
    private double fuel;

    public double getFuelPercentage() { return fuel; }
}

Car bmw = new Car();
bmw.getFuelPercentage();   // OK: read through a getter
```

Without encapsulation, anyone could write:

```java
bmw.gasType = "petrol";   // direct access, dangerous
```

This could break the engine logic, since a diesel car would suddenly be using petrol. With `private`, this line gives a compile error, so the class stays in control.

---

## 7. Testing

- All changes go through a few methods, so you test **those methods** only.
- If the inside changes later, the tests still pass as long as the methods behave the same.

---

## Summary

| Without encapsulation | With encapsulation |
|---|---|
| Anyone edits data directly | Data changes only through methods |
| No validation | Rules are checked in setters |
| Hard to track dependencies | Dependencies are handled inside the class |
| Hard to test | Easy to test |

# Abstraction

## 1. What is it?

Show **what** the output is, but hide **how** it was done. The user sees only the important parts, and the complicated details stay hidden.

```
User  ->  [ Simple Interface ]  ->  Result
              (complexity hidden behind it)
```

---

## 2. Examples

### Coffee machine
You put in water and choose the cup type, then press a button.
- **Goal:** drink coffee.
- **Hidden:** heating the water, pressure, grinding, timing.

### Car
You use the steering wheel and pedals, and that is all you need.
- **Hidden:** the engine, gearbox, fuel injection (the mechanical parts).

---

## 3. Interface vs Implementation

| | Meaning | Example (Car) |
|---|---|---|
| **Interface** | **What** can be done (the contract) | `start()`, `accelerate()`, `brake()` |
| **Implementation** | **How** it is done (the details) | Petrol engine vs electric motor |

```java
interface Vehicle {
    void start();
    void accelerate();
}

class PetrolCar implements Vehicle {
    public void start()      { /* ignite fuel, spin engine */ }
    public void accelerate() { /* open throttle */ }
}

class ElectricCar implements Vehicle {
    public void start()      { /* power the motor */ }
    public void accelerate() { /* increase current */ }
}
```

The driver calls `start()` the same way for both cars. How it works inside is different, but the driver does not care.

---

## 4. Example: Chess game

We want our class to **talk with different pieces through one interface**, without knowing their internal details.

```java
interface Piece {
    boolean isAlive();
    void move(int row, int col);
}

class Queen implements Piece {
    private boolean alive = true;
    public boolean isAlive() { return alive; }
    public void move(int row, int col) { /* queen movement rules */ }
}

class Alph implements Piece {   // Alph = Bishop (الفيل)
    private boolean alive = true;
    public boolean isAlive() { return alive; }
    public void move(int row, int col) { /* diagonal movement rules */ }
}
```

**Scenario:** the Alph and the Queen are on the same team. Before the Alph acts, it needs to know the Queen's **state** (alive or dead), and it only needs to call `isAlive()`.

```java
class Game {
    void checkTeam(Piece p) {
        if (p.isAlive()) {
            // piece can still move
        }
    }
}
```

`Game` works with **any** `Piece` (Queen, Alph, Knight...) and never needs to know how each one stores its state or moves. If a new piece is added later, `Game` does not change.

---

## 5. Why we need it

- **Simplicity:** the user deals with a small, clear interface.
- **Flexibility:** you can change the implementation without breaking the code that uses it.
- **Extensibility:** adding a new type (a new piece or car) is easy.
- **Less coupling:** classes depend on the interface, not on each other's details.

---

## Summary

| Concept | One line |
|---|---|
| **Abstraction** | Show **what**, hide **how** |
| **Interface** | The contract: what can be done |
| **Implementation** | The hidden details of how it is done |

---

## Encapsulation vs Abstraction (quick reminder)

- **Encapsulation** = hide the **data** (use `private` + getters/setters).
- **Abstraction** = hide the **complexity** (use `interface` / `abstract class`).
