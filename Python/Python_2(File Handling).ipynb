## File Handling ##

- "r" - Read - Default value. Opens a file for reading, error if the file does not exist

- "a" - Append - Opens a file for appending, creates the file if it does not exist

- "w" - Write - Opens a file for writing, creates the file if it does not exist

- "x" - Create - Creates the specified file, returns an error if the file exists

```python
f=open("Shop.txt",'r')
print(f) #handler
```

## Exception Names in Python

### 1. ImportError
**Meaning:** Python cannot import something from a module.

**Example:**







```python
from math import hello 
# Why ? the math module does not have something called hello
```

 ### 2. IndexError
**Meaning:** You try to get an item from a list using an index that does not exist.

**Example:**

```python
numbers = [10, 20, 30]
print(numbers[5])
#Why? The list has only 3 items. Their indexes are 0, 1, and 2. Index 5 does not exist.
```

### 3. NameError
**Meaning:** You use a variable name that Python does not know.

**Example:**

```python
print(age)
#Why? We did not create a variable called age before using it.
```


### 4. SyntaxError
**Meaning:** You write code with incorrect Python grammar.

**Example:**

```python
"""
if age > 18
    print("Adult")
    Why? We forgot to add a colon : after 18.
    """
#the correct code
if age > 18:
    print("Adult")
```

### 5. TypeError
**Meaning:** You try to do something with the wrong type of data.

**Example:**

```python
age = 20
print("My age is " + age)
#Why? "My age is " is a string, but age is an integer. Python cannot join them directly using +.
# the correct
print("My age is " + str(age))
```

### 6. ValueError
**Meaning:** You use the correct type of data, but its value is not acceptable for the operation.

**Example:**

```python
age = int("hello")
#Why? int() changes a value into an integer, but "hello" is not a number.
# the correct code
age = int("20")
print(age)
```

### Exception Handling ###
- try: Runs the risky code that might cause an error.
- except: Catches and handles the error if one occurs.
- else: Executes only if no exception occurs in try.
- finally: Runs regardless of what happens useful for cleanup tasks like closing files.

```python
try:
    print(10 / 0)
except:
    print("Something went wrong!")
```

```python
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Cannot divide by zero!")
else:
    print("The operation was successful!")
```

```python
try:
    a= int(input("please enter the first number:"))
    b= int(input("please enter the second number:"))
    print (b/a)
except Exception as h:
    print("Something is wrong")
print("all is good")
```

```python
try:
    a=7
    b=0
    print("S"+7)
    print(a/b)
    
except (ZeroDivisionError,TypeError):
    print("One of the operation is wrong  ")
```

```python
try :
    a=10
    print(10/0)

except ZeroDivisionError:
    print("Do not divide by zero")
except TypeError:
    print("Error Occured")
except NameError:
    print("Name Error")
```

### Else ###

```python
try:
    print(1)
except ZeroDivisionError:
    print(2)
else:
    print(3)
```

```python
try:
    print(1/0)
except ZeroDivisionError:
    print(2)
else:
    print(3)
print(4)    
```

### Finally ###

```python
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Cannot divide by zero!")
finally:
    print("The program has finished.")
```

### Raise Exception ###

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative.")
```
